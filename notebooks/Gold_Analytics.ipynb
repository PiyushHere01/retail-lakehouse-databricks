```python
# Loading th Silver Table.


silver_customers = spark.table("retail_project.silver_customers")

silver_sales = spark.table("retail_project.silver_sales")

silver_orders = spark.table("retail_project.silver_orders")
```

```python
display(silver_sales)
```

```python
# Total Sales KPI Logic. 
#  
from pyspark.sql.functions import sum

gold_total_sales = silver_sales.agg(sum("total_amount").alias("total_revenue"))

display(gold_total_sales)
```

```python
# Saving the Gold Table.

gold_total_sales.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("retail_project.gold_total_sales")
```

```python
# Creating Top Customer Ranking.

from pyspark.sql.functions import sum

gold_top_customers = silver_sales.groupBy("customer_name").agg(sum("total_amount").alias("total_spent")).orderBy("total_spent",ascending=False)

display(gold_top_customers)

# Saving the Table.

gold_top_customers.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("retail_project.gold_top_customers")
```

```python
# For Top Products.

gold_top_products = silver_sales.groupBy("product_name").agg(sum("total_amount").alias("product_revenue")).orderBy("product_revenue",ascending=False)

display(gold_top_products)

# Saving the Table.

gold_top_products.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("retail_project.gold_top_products")
```

```python
# Sales_Trend 

gold_sales_by_date = silver_sales.groupBy("order_date").agg(sum("total_amount").alias("daily_sales")).orderBy("order_date")

display(gold_sales_by_date)

# Saving the Table.

gold_sales_by_date.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("retail_project.gold_sales_by_date")
```
