```python
customers_df = spark.table(
    "databricks_simulated_retail_customer_data.v01.customers"
)

sales_df = spark.table(
    "databricks_simulated_retail_customer_data.v01.sales"
)

orders_df = spark.table(
    "databricks_simulated_retail_customer_data.v01.sales_orders"
)
```

```python
customers_df.count()
sales_df.count()
orders_df.count()
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS retail_project;
```

```python
customers_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("retail_project.bronze_customers")
```

```python
sales_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("retail_project.bronze_sales")
```

```python
orders_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("retail_project.bronze_orders")
```

```python
spark.sql("SHOW TABLES IN retail_project").show()
```

```python
spark.sql("""
DESCRIBE HISTORY retail_project.bronze_customers
""").show()
```
