```python
from pyspark.sql.functions import col

bronze_customers.filter(col("customer_id").isNull()).show()
```

```python
silver_customers = bronze_customers.dropDuplicates()
```

```python
silver_customers = silver_customers.dropna(subset=["customer_id"])
```

```python
from pyspark.sql.functions import upper, col

silver_customers = silver_customers.withColumn("customer_name",upper(col("customer_name")))
```

```python
silver_customers.count()
```

```python
silver_customers.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("retail_project.silver_customers")
```

```python
from pyspark.sql.functions import col, upper

# Load bronze sales table
bronze_sales = spark.table("retail_project.bronze_sales")

# Create silver sales dataframe
silver_sales = bronze_sales.dropDuplicates()

# Remove null customer IDs
silver_sales = silver_sales.dropna(subset=["customer_id"])

# Standardize product names
silver_sales = silver_sales.withColumn("product_name",upper(col("product_name")))

# Create total amount column
silver_sales = silver_sales.withColumn("total_amount",col("total_price"))

# Display final dataframe
display(silver_sales)
```

```python
from pyspark.sql.functions import col

# Remove duplicates
silver_orders = bronze_orders.dropDuplicates()

# Remove null order IDs
silver_orders = silver_orders.dropna(subset=["order_number"])

# Display cleaned data
display(silver_orders)
```

```python
# Validating the Count.
silver_orders.count()

# Saving the Silver Order Table.
silver_sales.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("retail_project.silver_sales")
```
