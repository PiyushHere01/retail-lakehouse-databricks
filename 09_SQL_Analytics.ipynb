```python
%sql

select product_name, sum(total_price) AS revenue
from retail_project.silver_sales
group by product_name
order by revenue desc
limit 10
```

```python
%sql

select order_date, sum(total_price) as daily_sales
from retail_project.silver_sales
group by order_date
order by order_date
```

```python
%sql

select product_category,
sum(total_price) as category_sales
from retail_project.silver_sales
group by product_category
order by category_sales desc
```
